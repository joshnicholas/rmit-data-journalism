```python
### Making a request

import requests 
import json 

response = requests.get('https://api.open-meteo.com/v1/forecast?latitude=-37.8136&longitude=144.9631&current_weather=true')

dicto = json.loads(response.text)['current_weather']
print('Temperature: ', dicto['temperature'])
```
