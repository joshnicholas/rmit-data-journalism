```python
import pandas as pd
from sudulunu.helpers import pp, dumper


data = pd.read_csv('data/population-with-un-projections.csv')
```

```python
df = data.copy()
# ['Entity', 'Code', 'Year', 'Population', 
#  'Population (Projected)']

keep = ['Australia', 'New Zealand', "Canada",'Chile','Cameroon',
        'Sri Lanka', 'Ecuador', 'Nepal', 'Mali']

df = df.loc[df['Entity'].isin(keep)]
# df = df.loc[df[]
df.dropna(subset=['Population'], inplace=True)
df = df.loc[df['Year'] == df['Year'].max()]

df = df[['Entity', 'Year', 'Population']]

df.rename(columns={'Entity': "Country"}, inplace=True)

df['Population'] = df['Population'].astype(int)

pp(df)

dumper("data", 'population_2023', df)

```
