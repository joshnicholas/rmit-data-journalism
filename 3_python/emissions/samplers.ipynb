```python
import pandas as pd 
from sudulunu.helpers import pp, dumper
```

```python
data = pd.read_csv('full_data/annual-co2-emissions-per-country/annual-co2-emissions-per-country.csv')
data.dropna(subset=['Code'], inplace=True)
data = data.loc[~data['Code'].str.contains("OWID")]

pp(data)
```

```python
oz = data.loc[data['Entity'] == 'Australia'].copy()
oz = oz.loc[oz['Annual CO₂ emissions'] > 0]

pp(oz)

dumper('emissions', 'aus_emissions_over_time', oz)
```

```python
latest = data.loc[data['Year'] == data['Year'].max()]

for i in range(0, 20):
    inter = latest.sample(n=5)

    dumper('emissions/small',str(i), inter)

# pp(latest)
```
